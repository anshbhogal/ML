```python
q1 = data["horsepower"].quantile(0.25)
q3 = data["horsepower"].quantile(0.75)
IQR = q3-q1
min_value = q1-(IQR*1.5)
max_value = q3+(IQR*1.5)
# data["horsepower"].apply() = data[(data["horsepower"] > max_value)].replace(max_value)
# data["horsepower"].apply() = data[(data["horsepower"] < min_value)].replace(min_value)

# data[(data["horsepower"] > max_value)]

data["horsepower"] = data["horsepower"].apply(lambda x: max_value if x > max_value else (min_value if x < min_value else x))
```
